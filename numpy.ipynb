Arithmetic opertion on numpy arrays

##### conditional operators

```python
a = np.array([10,20,30,40,50])
print(a>20)
print(a[a>20])
print(a>=20)
print(a[a>=20])
```

#### universal functions

```python
c = np.array([4,9,16,11])
print(np.sqrt(c))
print(np.square(c))
```

```python
d =np.array([-1,-2])
print(np.abs(d))
```

#### Aggregate function

```python
d = np.array([10,20,30,40])
print(np.sum(d))
print(d.sum())
print(np.max(d))
print(np.min(d))
print(np.mean(d))
print(np.median(d))
e = np.array([10,30,20,40,50])
print(np.median(e))
```

```python
f =np.array([10,20,30])
print(np.std(f))
print(np.var(f))
print(np.argmax(f))
print(np.argmin(f))
```

#### working with 2d arrays
 #### axis = 0 - column 
#### axis = 1 - row
#### axis = none - all over

```python
arr_2d=np.array([[1,2],[2,3],[4,5]])
print(np.sum(arr_2d,axis = 0))
print(np.sum(arr_2d,axis = 1))
print(np.sum(arr_2d,axis = None))
print(np.mean(arr_2d,axis =1))
```
