### Apply Function

define a function that returns
if marks greater than 80 return A
if marks greater than 60 return B
if marks greater than 40 return c
return fail

```python
def gradeDef(marks):
    if marks>80:
        return 'A'
    elif marks>60:
        return 'B'
    elif marks>40:
        return 'C'
    else:
        return 'F'
print(gradeDef(40))
```

```python
df['Grade']=df['marks'].apply(gradeDef)
df
```

```python
df['Bonus']=df['marks'].apply(lambda x:x+5)
df
```

### join in python (merge)

```python
students = pd.DataFrame({'student_id':[1,2,3],'name':['aishaa','asmath','adina']})
marks = pd.DataFrame({'student_id':[1,2,3,4],'marks':[100,50,20,90]})
```

```python
pd.merge(
    students,
    marks,
    on='student_id',
    how='inner'
)
```

```python
pd.merge(
    students,
    marks,
    on='student_id',
    how='left'
)
```

```python
pd.merge(
    students,
    marks,
    on='student_id',
    how='outer'
)
```

```python
s = pd.DataFrame({'student_id':[1,2,3],'name':['aishaa','asmath','adina']})
m = pd.DataFrame({'id':[1,2,3,4],'marks':[100,50,20,90]})
```

```python
pd.merge(
    s,
    m,
    left_on='student_id',
    right_on='id'
    how='left'
)
```
